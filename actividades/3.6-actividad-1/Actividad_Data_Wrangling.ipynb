La función `pd.concat()` permite combinar varios DataFrames. La concatenación

## Ejercicio 4.08: Combinación mediante una clave común

La función `pd.merge()` permite combinar DataFrames mediante una columna que 
funciona como clave común.

En este ejercicio se utilizan dos tipos de combinación:

- `inner`: conserva solamente las claves que existen en ambos DataFrames.
- `outer`: conserva todas las claves, aunque no tengan una coincidencia.

```python
df_1 = (df[["Ship Date", "Ship Mode", "Customer Name"]].iloc[0:4].copy())
df_1
```

```python
df_2 = (df[["Customer Name", "Product Name", "Quantity"]].iloc[0:4].copy())
df_2
```

```python
inside_union = pd.merge(df_1, df_2, on="Customer Name", how="inner")
print(f"Dimensions: {inside_union.shape}")
inside_union
```

```python
no_duplicates_inside_union = (inside_union.drop_duplicates().reset_index(drop=True))
print(f"Dimensions: {no_duplicates_inside_union.shape}")
no_duplicates_inside_union
```

```python
df_3 = (df[["Customer Name", "Product Name", "Quantity"]].iloc[2:6].copy())
df_3
```

```python
union_inner_13 = (pd.merge(df_1, df_3, on="Customer Name", how="inner").drop_duplicates().reset_index(drop=True))
print(f"Dimensions: {union_inner_13.shape}")
union_inner_13
```

```python
union_outer_13 = (pd.merge(df_1, df_3, on="Customer Name", how="outer").drop_duplicates().reset_index(drop=True))
print(f"Dimensions: {union_outer_13.shape}")
union_outer_13
```

```python
origin_union = (
    pd.merge(
        df_1,
        df_3,
        on="Customer Name",
        how="outer",
        indicator=True
    )
    .drop_duplicates()
    .reset_index(drop=True)
)

origin_union
```

### Interpretación

La combinación interna conservó solamente los clientes presentes en ambos 
DataFrames. Por esta razón, al combinar `df_1` y `df_3`, se conservaron los 
registros de Darrin Van Huff y Sean O'Donnell.

La combinación externa conservó todos los clientes. Cuando un cliente no tenía 
información correspondiente en la otra tabla, pandas agregó valores `NaN` o `NaT`.

### Conclusión

La función `pd.merge()` permite relacionar diferentes tablas mediante una clave 
común. La combinación `inner` resulta útil cuando solamente interesan los registros 
coincidentes, mientras que la combinación `outer` permite conservar toda la 
información disponible.

También se observó que una clave repetida puede generar varias combinaciones. 
Por ello, es importante revisar y eliminar duplicados cuando el contexto del análisis 
así lo requiera.

## Ejercicio 4.09: Método join()

El método `join()` combina las columnas de dos DataFrames utilizando sus índices. 
En este ejercicio, la columna `Customer Name` se convierte en el índice de ambas 
tablas y se aplican combinaciones `left`, `right`, `inner` y `outer`.

```python
df_1 = df[["Customer Name", "Ship Date", "Ship Mode"]].iloc[0:4].copy().set_index("Customer Name")
df_1
```

```python
df_2 = df[["Customer Name", "Product Name", "Quantity"]].iloc[2:6].copy().set_index("Customer Name")
df_2
```

```python
print("f_1 index:")
print(df_1.index.tolist())

print("\ndf_2 index:")
print(df_2.index.tolist())
```

```python
left_join = df_1.join(df_2, how="left").drop_duplicates()

print(f"Dimensions: {left_join.shape}")
left_join
```

```python
right_join = df_1.join(df_2, how="right").drop_duplicates()

print(f"Dimensions: {right_join.shape}")
right_join
```

```python
inner_join = df_1.join(df_2, how="inner").drop_duplicates()

print(f"Dimensions: {inner_join.shape}")
inner_join
```

```python
outer_join = df_1.join(df_2, how="outer").drop_duplicates()

print(f"Dimensions: {outer_join.shape}")
outer_join
```

### Interpretación

- `left` conservó todos los índices de `df_1`.
- `right` conservó todos los índices de `df_2`.
- `inner` conservó únicamente los índices presentes en ambos DataFrames.
- `outer` conservó todos los índices de las dos tablas.
- Los valores `NaN` y `NaT` indican que no se encontró información correspondiente.

### Conclusión

El método `join()` permitió combinar dos DataFrames utilizando `Customer Name` 
como índice. Cada tipo de unión produjo un resultado diferente según los registros 
que debía conservar.

El `inner join` solamente mantuvo las coincidencias, mientras que el `outer join` 
conservó todos los clientes. Los métodos `left` y `right` dieron prioridad al 
DataFrame ubicado a la izquierda o a la derecha, respectivamente.

A diferencia de `merge()`, que normalmente relaciona columnas, `join()` trabaja 
principalmente con los índices de los DataFrames.

## Ejercicio 4.10: Muestreo aleatorio

El método `sample()` permite seleccionar registros aleatorios de un DataFrame.

Sus parámetros principales son:

- `n`: cantidad exacta de registros.
- `frac`: proporción del conjunto de datos.

```python
sample_with_replacement = df.sample(frac=0.1, random_state=42, replace=True)
sample_with_replacement.head()
```

### Conclusión

El método `sample()` permitió obtener muestras aleatorias del conjunto de datos. 
El parámetro `n` seleccionó una cantidad exacta de registros, mientras que `frac` 
seleccionó una proporción del DataFrame.

Al utilizar `replace=False`, cada registro pudo aparecer solamente una vez. En 
cambio, con `replace=True`, algunos registros fueron seleccionados repetidamente. 
Finalmente, `random_state` hizo posible reproducir los mismos resultados cada vez 
que se ejecutó el notebook.

## Ejercicio 4.11: Ordenamiento con sort_values()

El método `sort_values()` permite ordenar los registros de un DataFrame utilizando 
una o varias columnas.

Por defecto, pandas ordena los valores de manera ascendente. Cuando se utilizan 
varias columnas, se aplica primero el orden de la primera y después el de las 
columnas siguientes.

```python
df_sample = df[["Customer Name", "State", "Sales", "Quantity"]].sample(n=15, random_state=42)
df_sample
```

```python
sorted_sales_sample = df_sample.sort_values(by="Sales")
sorted_sales_sample
```

```python
sorted_sales_state_sample = df_sample.sort_values(by=["State", "Sales"])
sorted_sales_state_sample
```

### Interpretación

Al ordenar únicamente por `Sales`, los registros se organizaron desde la venta 
más pequeña hasta la más grande.

Al ordenar por `State` y `Sales`, pandas organizó primero los estados 
alfabéticamente. Después ordenó las ventas correspondientes a cada estado. 
El parámetro `ascending` permitió controlar si el orden era ascendente o descendente.

### Conclusión

El método `sort_values()` permitió reorganizar la muestra mediante una o varias 
columnas. El ordenamiento por ventas facilitó la identificación de los valores 
mínimos y máximos, mientras que el ordenamiento por estado y ventas permitió 
organizar los registros jerárquicamente.

Ordenar los datos facilita su interpretación y ayuda a detectar tendencias, valores 
extremos y relaciones entre diferentes categorías.

## Ejercicio 4.12: Funciones personalizadas con apply()

El método `apply()` permite ejecutar una función sobre cada elemento de una 
columna. En este ejercicio se utiliza para:

- Clasificar las ventas como bajas, medias o altas.
- Calcular la longitud del nombre de cada cliente.
- Aplicar un descuento mediante una función lambda.

```python
def categorize_sales(price):
    if price < 50:
        return "Low"
    elif price < 200:
        return "Medium"
    else:
        return "High"
```

```python
df_sample = df[["Customer Name", "State", "Sales"]].sample(n=100, random_state=42)
df_sample.head(10)
```

```python
df_sample["Sales Price Category"] = df_sample["Sales"].apply(categorize_sales)
df_sample.head(10)
```

```python
df_sample["Customer Name Lenght"] = df_sample["Customer Name"].apply(len)
df_sample.head(10)
```

```python
df_sample["Discounted Price"] = df_sample["Sales"].apply(lambda price: price * 0.85 if price > 200 else price)
df_sample.head(10)
```

### Interpretación

La función `categorize_sales()` transformó los valores numéricos de `Sales` en 
categorías de texto. Esto permite interpretar las ventas como bajas, medias o altas.

La función incorporada `len` calculó la longitud de los nombres de los clientes. 
Finalmente, una función lambda aplicó un descuento del 15 % únicamente a las 
ventas mayores que 200.

### Conclusión

El método `apply()` permitió ejecutar diferentes funciones sobre los elementos de 
las columnas del DataFrame. Se utilizó una función personalizada para clasificar las 
ventas, una función incorporada para calcular la longitud de los nombres y una 
función lambda para aplicar descuentos condicionales.

Esta herramienta proporciona flexibilidad para crear nuevas variables, transformar 
datos y aplicar reglas de negocio durante el proceso de Data Wrangling.
